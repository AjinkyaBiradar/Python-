```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('IPL.csv')
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
matchwin = df['match_winner'].value_counts()
sns.barplot(y = matchwin.index , x= matchwin.values)
plt.title('Most Match Win BY TEam')
```

```python
sns.countplot(x = df['toss_decision'])
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count *100)/df.shape[0]
percentage.round(2)
```

```python
sns.countplot(x =df['won_by'])
plt.title('Won by')
```

```python
count = df['player_of_the_match'].value_counts().head(10)
```

```python
sns.barplot(x = count.values, y = count.index, palette = 'mako')
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending= False).head(3)
high
```

```python
high.plot(kind = 'barh')
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending = False).head(7)
top_bowlers.plot(kind = 'barh')
```

```python
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(y = venue_count,x = venue_count.values, palette ='mako')
```
